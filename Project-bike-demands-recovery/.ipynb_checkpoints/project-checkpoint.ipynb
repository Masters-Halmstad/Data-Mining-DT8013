# Some initial thoughts
- we have 165 missing hours
- while cnt is accurate for the given data but the smallest hourly cnt is 1 which is interesting since there must be a time where there are no riders,

```python
# investigating missing hours 
full_index_hour = pd.date_range(df_hour.timestamp.min(), df_hour.timestamp.max(), freq='h')
missing = full_index_hour.difference(df_hour.timestamp).sort_values()
missing
```

```python
# grouping subsequent gaps into episode run
run_id = (pd.Series(missing).diff() != pd.Timedelta('1h')).cumsum().values
gaps = (pd.DataFrame({'ts': missing, 'run': run_id})
        .groupby('run').ts.agg(start='min', end='max', hours='size'))

print('gap length (h) -> number of gaps :', gaps.hours.value_counts().sort_index().to_dict())
print('missing hours by hour of day      :', pd.Series(missing.hour).value_counts().sort_index().to_dict())
gaps.sort_values('hours', ascending=False).head(10)
```

```python
short = gaps[gaps.hours <= 2]

# count short gaps by the hour they start, keeping all 24 hours so empty ones show as 0
freq = short.start.dt.hour.value_counts().reindex(range(24), fill_value=0)
fig, ax = plt.subplots(figsize=(10, 4))
bars = ax.bar(freq.index, freq.values, width=0.7, color='#2a78d6')
ax.bar_label(bars, labels=[str(v) if v else '' for v in freq.values],
             padding=2, fontsize=9, color='#52514e')

ax.set_xticks(range(24))
ax.set_xlabel('hour of day the gap starts')
ax.set_ylabel('number of gaps')
ax.set_title(f'Short gaps (≤ 2 h) by start hour (n = {len(short)})', loc='left')

ax.grid(axis='y', color='#e5e4e0', linewidth=0.8)
ax.set_axisbelow(True)
for side in ('top', 'right'):
    ax.spines[side].set_visible(False)

plt.tight_layout()
plt.show()
```

## Our assumption

### 70 short gaps
- there are alot of short gaps 70 1-2h short gaps, almost all between 02:00 to 05:00, we are thinking it's because there are no rider at particular time so the system just doesn't register any values at all, which also explains why we have no 0 in the 'cnt'  column.

### The big gaps (Confirmed through web search)
- 36h gap between 29-30 Oct 2012 was caused by Hurricane Sandy (in data readme as well)
- 22h in January 26–27, 2011: Crippling Evening Commute Snowstorm
- 13h in August 27–28, 2011: Hurricane Irene
- 12h in January 18, 2011: Freezing Rain and Ice
- 6h in February 22, 2011: Overnight Snow and Sleet

All the short as well as big gaps are explained. Although, the night gaps (the short gaps) are because there was no demand but the longer gaps are due to weather events which forced the Capital Bikeshare itself to be closed down, meaning supply was zero. We can treat the 5 events as anomalies events.

Since we are trying to add a new feature called "hours_after" which will track how many hour it has been after rain. So we need to fill those missing rows so we have a continues timeseries.


Lag features like "rentals 1 hour ago" or "same hour yesterday", and rolling averages, assume each row is exactly one hour after the previous one. With rows missing, "the previous row" is sometimes 2 hours or 36 hours ago.
Missing hours aren't all the same. Your gap analysis showed two kinds, and each needs a different fix:
- Short gaps at night most likely mean nobody rented a bike, so the real value is 0.
- Long gaps (Sandy, Irene, the snowstorm) mean the service was probably shut down. The real demand is unknown, not 0. Filling those with 0 would teach a model that a Monday afternoon in October can have zero riders.
- So we are going to add another column called outages, any gap more than 5 hour is treated as an outage.
- 0 means nobody wanted to rent a bike which is different than having unable to rent a bike. If we fill outage hour with 0, a model will learn that a weekday afternoon can have zoer riders, which is fales and will pull prediction down. We'll keep them as NaN with an outage flag that will let up drop those gaps before training a model so it only learns from real observations and still see smth happened. The model can learn that storms reduce demand from the observered hours around them.

```python
# added outage flag and filling in non outages missing hours/rows with 0 riders
outage_threshold = 5
outage_hours = pd.DatetimeIndex(np.concatenate(
    [pd.date_range(g.start, g.end, freq='h') for g in gaps.itertuples() if g.hours >= outage_threshold]))
# adding missing rows
df = df_hour.set_index("timestamp").reindex(full_index_hour)
# outage for long gaps
df['outage'] = df.index.isin(outage_hours)
# missing but not an outage gets zero
df['zero_fill'] = df['cnt'].isna() & ~df['outage']
# calender column
df['dteday'] = df.index.normalize()
df['hr'] = df.index.hour
df_day['dteday'] = pd.to_datetime(df_day['dteday'])
cal = df_day.set_index('dteday')[['season', 'yr', 'mnth', 'holiday', 'weekday', 'workingday']]
df[cal.columns] = cal.reindex(df['dteday']).values.astype(int)

# filling zero
df.loc[df.zero_fill, ['casual', 'registered', 'cnt']] = 0
# since weather changes slowly so missing hours copy the last observed hour (ffill)
df[weather_cols] = df[weather_cols].ffill()

df
print(f'zero-rental hours filled with 0: {df.zero_fill.sum()} | outage hours kept as NaN: {df.outage.sum()} | hours: {len(df)} hours')
```

Since our data rows are complete timeseries, let's check if all the sensors values are valid or not.

```python
df['temp_c']   = df['temp'] * 41        # "divided to 41 (max)"
df['atemp_c']  = df['atemp'] * 50       # "divided to 50 (max)"
df['hum_pct']  = df['hum'] * 100        # "divided to 100 (max)"
df['wind_kmh'] = df['windspeed'] * 67   # "divided to 67 (max)"

WEATHER = {1: 'clear', 2: 'mist/cloudy', 3: 'light rain/snow', 4: 'heavy rain/snow'}
SEASON  = {1: 'spring', 2: 'summer', 3: 'fall', 4: 'winter'}
# ffill turned it into float
df['weathersit']   = df['weathersit'].astype(int) 
df['weather']      = df['weathersit'].map(WEATHER)
df['season_name']  = df['season'].map(SEASON)
df['date']         = df.index.date
df['weekday_name'] = df.index.day_name()

df[['date', 'hr', 'weekday_name', 'season_name', 'weather',
    'temp_c', 'atemp_c', 'hum_pct', 'wind_kmh', 'cnt']].head()
```

## The inconsistency 
If we follow the readme direction for conversion of weather column values, it gives us unrealistic temp, 9.02 degree in heavy rain / snow in DC and no hour where temp didnot fall below 0 ? where there is heavy snow ? is unrealistic. 

We found the original dataset https://archive.ics.uci.edu/dataset/275/bike+sharing+dataset
in which the conversion formula is 
- temp : Normalized temperature in Celsius. The values are derived via (t-t_min)/(t_max-t_min), t_min=-8, t_max=+39 (only in hourly scale)
- atemp: Normalized feeling temperature in Celsius. The values are derived via (t-t_min)/(t_max-t_min), t_min=-16, t_max=+50 (only in hourly scale)
   

```python
# UCI formula: value = (t - t_min) / (t_max - t_min)  ->  t = t_min + value * (t_max - t_min)
df['temp_c']  = -8  + df['temp']  * (39 - (-8))     # t_min = -8,  t_max = 39  -> -8 + 47·temp
df['atemp_c'] = -16 + df['atemp'] * (50 - (-16))    # t_min = -16, t_max = 50  -> -16 + 66·atemp
print(df[['temp_c', 'atemp_c']].agg(['min', 'max']).round(1))
print('hours below 0 °C:', (df.temp_c < 0).sum())
print("-"*50)
# 2011-01-26 known weather anomoly, snow strome

df[weather_cols + ['outage','weather','season_name','temp_c', 'atemp_c']].sort_values('temp_c').head(10)
```

## we cannot have humidity 0, otherwise model will learn that rain/snow can happen at hum 0 which is not true.

```python
bad_hum = df.hum_pct.eq(0)
df['hum_filled'] = bad_hum
# typical humidity per (month, weather type),
# computed only from good, observed hours
good = df[~bad_hum & ~df.outage]
hum_med = good.groupby(['mnth', 'weathersit'])['hum_pct'].median()

# look up the median for each bad row's (month, weathersit) pair
keys = pd.MultiIndex.from_frame(df.loc[bad_hum, ['mnth', 'weathersit']])
df.loc[bad_hum, 'hum_pct'] = hum_med.reindex(keys).values
# keep the normalised column consistent
df.loc[bad_hum, 'hum'] = df.loc[bad_hum, 'hum_pct'] / 100

print('0 % hours left:', (df.hum_pct == 0).sum())
df.loc['2011-03-10', ['hr', 'weather', 'hum_pct', 'hum_filled']].head(8)
```

## Checking for stuck or broken sensor.

```python
# stuck sensor the same value repeated for many hours in a row.
# hours we filled ourselves are skipped, because ffill / median made them flat on purpose
sensors = ['temp_c', 'atemp_c', 'hum_pct', 'wind_kmh']
filled = df.zero_fill | df.outage | df.hum_filled

for col in sensors:
    s = df[col].where(~filled)
    # new id every time the value changes
    run_id = (s != s.shift()).cumsum()
    runs = s.groupby(run_id).agg(start=lambda x: x.index[0], hours='size', value='first').dropna()
    print(f"\n{col}: 99.9% of runs are <= {runs.hours.quantile(0.999):.0f} h | longest runs:")
    print(runs.nlargest(4, 'hours').to_string(index=False))

```

### Our interpertataion
- atemp is frozen at −0.0016 °C for 24 h on 2012-08-17 and 2012-12-29 while temp changes during the day, so atemp is stuck on those days.
- for 2011-09-06 both temp_c and atemp_c gives the same value for 25 hour so it's correlated so we are trusting it.

- and for wind_kmh, we are checking the validity of the value, if it's a default value when there is no reading or actual value when there low/no wind.

```python
# wind speeds near 0: is 0 a real reading or "too calm to measure"?
wind = df.loc[~filled, 'wind_kmh'].round()
counts = wind[wind <= 20].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(counts.index, counts.values, width=0.7, color='#2a78d6')
ax.set_ylim(0, counts.max() * 1.12)

ax.set_xticks(range(0, 21, 2))
ax.set_xlabel('wind speed (km/h)')
ax.set_ylabel('number of hours')
ax.set_title('Wind speeds', loc='left')
ax.grid(axis='y', color='#e5e4e0', linewidth=0.8)
ax.set_axisbelow(True)
for side in ('top', 'right'):
    ax.spines[side].set_visible(False)
plt.tight_layout()
plt.show()
```

## Our Interpretation
- 2,180 hours (about 12.5%) are exactly 0 km/h, and there are no readings between 1 and 5 km/h.
- Above 6 km/h the readings come every 1–2 km/h (6, 7, 9, 11, 13 …). If the sensor could measure light wind, we would expect bars at 2 and 4 km/h as well.
- So 0 means "below what the sensor can measure". It isn't a missing value or a broken sensor.
- The long 0 km/h runs all fall overnight when wind usually dies down, so they're calm nights. We keep all 0 km/h values.

## Fixing the stuck temp
- atemp should be highly correlated with temp itself which can be used to develop a predictive model to determine the best approximate value for the missing one (lecture 2, slide 31)

```python
# a day is stuck when atemp has one value all day while temp changes
per_day = df.groupby('dteday').agg(atemp_values=('atemp_c', 'nunique'),
                                   temp_values=('temp_c', 'nunique'))
# based on temp changes and atemp not changing we find the stuck days
stuck_days = per_day.index[(per_day.atemp_values == 1) & (per_day.temp_values > 1)]
per_day.loc[stuck_days]
```

```python
# atemp follows temp closely, so fit a straight line on the good hours
# filtering out temps that we filled ourself
stuck = df.dteday.isin(stuck_days)
# OR keeps the flags if this cell is re-run after the fix
df['atemp_filled'] = df.get('atemp_filled', False) | stuck

# atemp follows temp closely, so fit a straight line on the good hours
# (not the hours we filled ourselves, and not the stuck hours)
good = ~filled & ~df.atemp_filled

coef = np.polyfit(df.loc[good, 'temp_c'], df.loc[good, 'atemp_c'], 1)
print(f"atemp ≈ {coef[0]:.2f} × temp {coef[1]:+.2f} | corr on good hours: "
      f"{df.loc[good, 'temp_c'].corr(df.loc[good, 'atemp_c']):.3f}")

df.loc[df.atemp_filled, 'atemp_c'] = np.polyval(coef, df.loc[df.atemp_filled, 'temp_c'])
# keeping the normalised column consistent
df.loc[df.atemp_filled, 'atemp'] = (df.loc[df.atemp_filled, 'atemp_c'] + 16) / 66

df.loc['2012-08-17', ['hr', 'temp_c', 'atemp_c', 'atemp_filled']].head(8)
```

## New columns added to the dataset

**Flags: which rows we added or changed** (True/False)
| Column | Meaning |
|---|---|
| `outage` | Hour inside one of the 5 long gaps (5 h or longer) caused by storm shutdowns. Ride counts are left empty (NaN), because demand is unknown, not 0. |
| `zero_fill` | Hour missing in a short gap (1–2 h, mostly at night). Ride counts were filled with 0, because nobody rented a bike. |
| `hum_filled` | Humidity was 0 % (2011-03-10, 24 h) and was replaced with the median humidity for that month and weather type. |
| `atemp_filled` | atemp was stuck at one value all day (2012-08-17 and 2012-12-29, 48 h) and was replaced with a prediction from temp. |

**Weather in real units** (converted from the normalised 0–1 columns)
| Column | Meaning |
|---|---|
| `temp_c` | Temperature in °C, using the UCI formula: −8 + 47 × temp |
| `atemp_c` | "Feels like" temperature in °C, using the UCI formula: −16 + 66 × atemp |
| `hum_pct` | Relative humidity in % (hum × 100) |
| `wind_kmh` | Wind speed in km/h (windspeed × 67). 0 means calm, below what the sensor measures. |

**Readable labels** (for plots and tables)
| Column | Meaning |
|---|---|
| `weather` | `weathersit` as text: clear, mist/cloudy, light rain/snow, heavy rain/snow |
| `season_name` | `season` as text after our correction: 1 = winter, 2 = spring, 3 = summer, 4 = fall |
| `date` | Calendar date without the time |
| `weekday_name` | Day of the week as text (Monday, Tuesday, …) |

The table index is now `timestamp` (date + hour), with one row for every hour from 2011-01-01 00:00 to 2012-12-31 23:00 (17,544 rows).
