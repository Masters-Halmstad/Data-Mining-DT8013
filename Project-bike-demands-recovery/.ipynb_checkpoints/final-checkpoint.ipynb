## Research Question

### Does Bike Demand Stay Low After Bad Weather Ends? And if yes how long does it take for the demand to come back to normal? 

This project measures how long it takes for bike demand to return to normal after rain or snow stops and whether commuters (registered riders) and casual riders recover at different speeds. And how long a pattern lasts is one of the things that makes temporal data interesting (Lecture 7 – Slide 7). Our research question looks at what happened in the series rather than forecasting what will happen so it falls under time series mining (Lecture 6 – Slide 8). We build a baseline that follows the trend and seasonality of demand (Lecture 6 – Slides 16–19) and divide actual demand by it. This removes both and leaves how far demand falls during and after rain (Lecture 6 – Slides 20–21).

We use the hourly data (`hour.csv`), not the daily data because recovery after rain happens within hours and a daily total would mix a rainy morning with a dry afternoon. Our cleaning is driven by the recovery clock `hours_after_rain` which counts rows as hours so every hour must have exactly one row (regular measurements, Lab 4 – Task 1).

### The big gaps (Confirmed through LLM web search (Lecture 7 – Slide 62) Automated Verification)
- 36h gap between 29-30 Oct 2012 was caused by Hurricane Sandy (in data readme as well)
- 22h in January 26–27, 2011: Crippling Evening Commute Snowstorm

- 12h in January 18, 2011: Freezing Rain and Ice

The short gaps are because there was no demand but the longer gaps are due to weather events which forced the Capital Bikeshare itself to be closed down meaning supply was zero. We can treat the 5 events as anomalies events, since they were generated by a different mechanism (Lecture 4 – Slide 3).

Since we are trying to add a new feature called "hours_after_rain" which will track how many hour it has been after rain. So we need to fill those missing rows so we have a continues timeseries.


Lag features like "rentals 1 hour ago" or "same hour yesterday" and rolling averages, assume each row is exactly one hour after the previous one. With rows missing, "the previous row" is sometimes 2 hours or 36 hours ago.
- Short gaps at night most likely mean nobody rented a bike, so the real value is 0 and we define that value (Lecture 2 – Slide 31).
- Long gaps (Sandy, Irene, the snowstorm) mean the service was probably shut down. The real demand is unknown because there was no supply, so we leave these observations out (Lecture 2 – Slide 31).
- We are going to add another column called outages and any gap more than 5 hour is treated as an outage which can be easily filtered out.
- If we fill outage hour with 0, a model will learn that a weekday afternoon can have zero riders which is fales and will pull prediction down. We'll keep them as NaN with an outage flag that will let up drop those gaps before training a model so it only learns from real observations and still see something happened. The model can learn that storms reduce demand from the observered hours around them.

````python
# added outage flag and filling in non outages missing hours/rows with 0 riders
# and since outage was caused by extrem weather we fill weathersit as 4 
outage_threshold = 5
outage_hours = pd.DatetimeIndex(np.concatenate(
    [pd.date_range(g.start, g.end, freq='h') for g in gaps.itertuples() if g.hours >= outage_threshold]))
# adding missing rows
df = df_hour.set_index("timestamp").reindex(full_index_hour)
# outage for long gaps
df['outage'] = df.index.isin(outage_hours)
df.loc[df['outage'], 'weathersit'] = 4
# missing but not an outage gets zero
df['zero_fill'] = df['cnt'].isna() & ~df['outage']
# calender column
df['dteday'] = df.index.normalize()
df['hr'] = df.index.hour
# we pull day info for those missing hours from day dataset
df_day['dteday'] = pd.to_datetime(df_day['dteday'])
cal = df_day.set_index('dteday')[['season', 'yr', 'mnth', 'holiday', 'weekday', 'workingday']]
df[cal.columns] = cal.reindex(df['dteday']).values.astype(int)

# filling zero
df.loc[df.zero_fill, ['casual', 'registered', 'cnt']] = 0
# since weather changes slowly so missing hours copy the last observed hour (ffill)
df[weather_cols] = df[weather_cols].ffill()

print(f'zero-rental hours filled with 0: {df.zero_fill.sum()} | outage hours kept as NaN: {df.outage.sum()} | hours: {len(df)} hours')
df[df['outage'] == True].head()
# df[df['zero_fill'] == True].head()
````

Since our data rows are now complete timeseries, let's check if all the sensors values are valid or not (Lecture 1 – Slide 42).

````python
df['temp_c']   = df['temp'] * 41        # "divided to 41 (max)"
df['atemp_c']  = df['atemp'] * 50       # "divided to 50 (max)"
df['hum_pct']  = df['hum'] * 100        # "divided to 100 (max)"
df['wind_kmh'] = df['windspeed'] * 67   # "divided to 67 (max)"

WEATHER = {1: 'clear', 2: 'mist/cloudy', 3: 'light rain/snow', 4: 'heavy rain/snow'}
SEASON  = {1: 'spring', 2: 'summer', 3: 'fall', 4: 'winter'}
# ffill turned it into float
df['weathersit']   = df['weathersit'].astype(int) 
df['weather']      = df['weathersit'].map(WEATHER)
df['season_name']  = df['season'].map(SEASON)
df['date']         = df.index.date
df['weekday_name'] = df.index.day_name()

df[['date', 'hr', 'weekday_name', 'season_name', 'weather',
    'temp_c', 'atemp_c', 'hum_pct', 'wind_kmh', 'cnt']].head()
````

## The inconsistency 
If we follow the readme direction for conversion of weather column values, it gives us unrealistic temp, 9.02 degree in heavy rain / snow in DC and no hour where temp didnot fall below 0 ? This is unrealistic and is confirmed using our domain expert (DeepseekLLM). The metadata must be right for the data to be interpreted correctly (Lecture 2 – Slide 24).

We found the original dataset https://archive.ics.uci.edu/dataset/275/bike+sharing+dataset
in which the conversion formula is 
- temp : Normalized temperature in Celsius. The values are derived via (t-t_min)/(t_max-t_min), t_min=-8, t_max=+39 (only in hourly scale)
- atemp: Normalized feeling temperature in Celsius. The values are derived via (t-t_min)/(t_max-t_min), t_min=-16, t_max=+50 (only in hourly scale)
   

````python
# UCI formula: value = (t - t_min) / (t_max - t_min)  ->  t = t_min + value * (t_max - t_min)
df['temp_c']  = -8  + df['temp']  * (39 - (-8))     # t_min = -8,  t_max = 39  -> -8 + 47·temp
df['atemp_c'] = -16 + df['atemp'] * (50 - (-16))    # t_min = -16, t_max = 50  -> -16 + 66·atemp
print(df[['temp_c', 'atemp_c']].agg(['min', 'max']).round(1))
print('hours below 0 °C:', (df.temp_c < 0).sum())
print("-"*50)
# 2011-01-26 known weather anomoly, snow strome

df[weather_cols + ['outage','weather','season_name','temp_c', 'atemp_c']].sort_values('temp_c').head(10)
````

# Anomaly detection with kNN and LOF

 Task 1 asks us to find anomalies with different unsupervised methods and to justify them with external information. Unsupervised methods need no labels and can find anomalies. During our initial data cleaning process we might have already found 5 weather anomalies that caused the whole system to go down and are now marked as outage in the dataframe. 

- 36h gap between 29-30 Oct 2012 was caused by Hurricane Sandy (in data readme as well)
- 22h in January 26–27, 2011: Crippling Evening Commute Snowstorm
- 13h in August 27–28, 2011: Hurricane Irene
- 12h in January 18, 2011: Freezing Rain and Ice
- 6h in February 22, 2011: Overnight Snow and Sleet

 We use the two neighbourhood methods:
- kNN: a day is anomalous when it is far from its k nearest neighbours (Lecture 4 – Slide 53).
- LOF: a day is anomalous when its neighbourhood is much less dense than its neighbours' neighbourhoods, so it also finds days that are only odd locally (Lecture 4 – Slides 65).

Both are multivariate (Lecture 4 – Slides 37–40; Lab 2 – Task 2). There is no single universal algorithm (Lecture 5 – Slide 109) so we use both and trust the days that both rank high, as in Lab 3 – Task 6.


## The process
We run kNN and LOF in two passes with each on its own set of features. We are using feature bagging to find diverse type of outliers (Lecture 5 – Slides 86, 94). We pick the subsets by meaning instead of at random because the anomalies we are looking for come from different mechanisms:

- **Pass A (weather only)** finds weather anomalies: storms (meaningful anomalies) and broken sensors (noise, Lecture 4 – Slide 18).
- **Pass B (riders only)** finds unusual demand, whatever the cause is (holidays, events or storms).

Splitting the features gives us two things:
- **More contrast.** A sensor fault shows in one or two weather columns while the rider counts stay normal. Every normal column we add makes the faulty day look more like the others (curse of dimensionality, Lecture 5 – Slide 16), so each pass keeps only the features where its anomalies show (*"Feature selection is key!"*, Lecture 1 – Slide 22).
- **A clear cause.** A day flagged in Pass A is about the weather or a sensor. A day flagged in Pass B is about the riders and we check its weather and outside sources afterwards.


### Choices and reasoing: 


- **using days instead of hour timestep:**
    Weather and demand are time series so each hour is strongly tied to the hours around it. If we gave the detectors single hours the faulty hours of one day would be each other's nearest neighbours and hide (masking) and the detectors would mostly see the daily rhythm (rush hour vs night) instead of anomalies. Summarising each day into one row changes a bad day into one unusual point (one value per day as in Lab 4 – Task 1).

- **`n_neighbors : 15`**
   k being how many neighbours each day is compared with so it also decides how big a group of similar days can get before the detectors stop seeing it as unusual. **Our assumtion** is if a day has about 15 days just like it in two years it should be a normal (pattern) and we treat it as a normal pattern in the data not an anomaly.

## Pass A: weather sensors

We only use **observed hours**. The hours we filled ourselves (`outage`, `zero_fill`) are flat copies of the previous hour and would pollute the data. Each day is described by sensor features (Lecture 1 – Slide 43) using: the standard deviation, mean and peak of a reading over the day and how two related sensors agree (atemp vs temp).

````python
from pyod.models.knn import KNN
from pyod.models.lof import LOF
from pyod.utils.utility import standardizer

# only real observations
filled = df.outage | df.zero_fill
obs = df[~filled]
g = obs.groupby('dteday')

# we use std to check sensor reading changes
# no mean/max for temp and atemp, because those mostly follow the season
days = pd.DataFrame({f'{c}_std': g[c].std() for c in ['temp_c', 'atemp_c', 'hum_pct', 'wind_kmh']})
# atemp should follow temp so a large gap means one of them is wrong
days['feels_gap'] = g.atemp_c.mean() - g.temp_c.mean()
# humidity's level doesn't follow the season like temperature does so the mean
# tells us how wet the day was and the std above covers how much it changed
days['hum_mean'] = g.hum_pct.mean()
# wind is mostly calm with short strong bursts since the calm hours would pull the mean down
# and hide the peak so we keep the strongest hour of the day
days['wind_max'] = g.wind_kmh.max()
# weathersit is a category, so instead of averaging the codes we count how much of the day it rained or snowed
days['rain_share'] = g.weathersit.apply(lambda s: (s >= 3).mean())
days = days.dropna()

X = standardizer(days.values)
X
````

````python
# helper function to find the cut off for investigating anomalies using decision_scores_
def plot_knn_lof(knn, lof, fig_name):
    fig, axes = plt.subplots(1, 2, figsize=(11, 4))
    for ax, (name, scores) in zip(axes, [('kNN', knn), ('LOF', lof)]):
        s = np.sort(scores)[::-1][:40]
        ax.plot(range(1, len(s) + 1), s, marker='o', markersize=4, color='#2a78d6')
        ax.set_yscale('log')
        ax.set_ylabel(f'{name} score (log scale)')
        ax.set_title(f'{name}, top 40 of {len(scores)} days', loc='left')
        ax.grid(axis='y', color='#e5e4e0', linewidth=0.8)
        ax.set_axisbelow(True)
        for side in ('top', 'right'):
            ax.spines[side].set_visible(False)
    
    plt.tight_layout()
    savefig(fig_name)
    plt.show()
````

````python
knn_a = KNN(n_neighbors=15).fit(X).decision_scores_
lof_a = LOF(n_neighbors=15).fit(X).decision_scores_
plot_knn_lof(knn_a, lof_a, 'scores_weather')
````

### choice and reasoning

- **Standardising the features.** kNN and LOF measure distances and our features are in different units (°C, %, km/h). Without scaling the feature with the biggest numbers would decide the distance.
- **Combining kNN and LOF by rank.** The two scores are on different scales so averaging the scores isn't ideal. Rank is a standard output of unsupervised detectors and puts both on the same scale. Combining several detectors is the ensemble idea (Lecture 5 – Slide 94) and we averaged the detectors' ranks for the same reason in Lab 2.
- **The top 15 days.** We use the top-N output. In the score plots the kNN curves are almost flat after about rank 15 so the days after that are not clearly more unusual than the rest. 15 days is about 2 % of the 730 days, few enough to check each one against weather records and online sources.


````python
cut_off = 15
days['knn_rank'] = pd.Series(knn_a, index=days.index).rank(ascending=False).astype(int)
days['lof_rank'] = pd.Series(lof_a, index=days.index).rank(ascending=False).astype(int)

top = days.assign(mean_rank=days[['knn_rank', 'lof_rank']].mean(axis=1)).nsmallest(cut_off, 'mean_rank')
print('typical day:', days.median().round(1).to_dict())
top.round(2)
````

## Pass B: demand

### Choices and reasoning: 

- **Ratio for demand and median as primary representation of demand**:

    Assuming the system grew from 2011 to 2012 and summer is obviously busier than winter so using only raw counts would just flag winter. We divide each day by the median day of the same year, month and day type (working day or weekend/holiday) so 1.0 = a normal day. Each month of each year has its own median which removes the growth and the season and the day type removes the difference between working days and days off. After the trend and seasonality are compensated for the anomaly part remains (Lecture 6 – Slides 18–21). We use the median so the anomaly themselves don't pull the "normal day" down.
  
- **Using log to compensate for ratio loss of information**

    A ratio can't go below 0 but has no upper limit so all low days are squeezed between 0 and 1 while high days stretch up creating an uneven spread of data. We take the log of the ratio to this as this is the standard fix for an uneven spread (Lecture 6 – Slides 30–31; Lab 4 – Task 3). It works here because every day has at least a few rides.

````python
# in Pass B we are looking for anomalies in DEMAND. A day is only "low" compared with what is normal for that
# time of year, so divide each day by the median day of the same year, month and day type
daily_rides = df.groupby('dteday').agg(
    registered=('registered', 'sum'), casual=('casual', 'sum'),
    yr=('yr', 'first'), mnth=('mnth', 'first'), workingday=('workingday', 'first'),
    holiday=('holiday', 'first'), outage_hours=('outage', 'sum'),
    rain_hours=('weathersit', lambda s: (s >= 3).sum()))
print("Zero values ?\n", (daily_rides[['registered', 'casual']]==0).sum())

group = daily_rides.groupby(['yr', 'mnth', 'workingday'])

for rider in ['registered', 'casual']:
    daily_rides[rider + '_ratio'] = daily_rides[rider] / group[rider].transform('median')
    daily_rides[rider + '_log'] = np.log(daily_rides[rider + '_ratio'])
Xr = standardizer(daily_rides[['registered_log', 'casual_log']].values)
Xr
````

````python
knn_b = KNN(n_neighbors=15).fit(Xr).decision_scores_
lof_b = LOF(n_neighbors=15).fit(Xr).decision_scores_
plot_knn_lof(knn_b, lof_b, 'scores_demand')
````

### choice and reasoning
- **Standardising both log ratios.** Casual demand swings much more than registered demand so without scaling the casual column alone would decide the distance. Scaling gives both rider types the same weight.
- **Same k and cut-off as Pass A**, so the two passes are compared on equal terms.


````python
cut_off = 15
daily_rides['knn_rank'] = pd.Series(knn_b, index=daily_rides.index).rank(ascending=False).astype(int)
daily_rides['lof_rank'] = pd.Series(lof_b, index=daily_rides.index).rank(ascending=False).astype(int)

top_demand = daily_rides.assign(mean_rank=daily_rides[['knn_rank', 'lof_rank']].mean(axis=1)).nsmallest(cut_off, 'mean_rank')
top_demand[['registered_ratio', 'casual_ratio', 'rain_hours', 'outage_hours', 'holiday', 'knn_rank', 'lof_rank']].round(2)
````

# Fixing what the detectors found

Pass A found three sensor faults. Each fix uses one of the approaches for missing values in Lecture 2 – Slide 31, and every changed hour gets a flag column.

## Humidity 0 % on 2011-03-10

Humidity has no strong partner variable to predict it from, so we use the typical value for that month and weather type. The median is taken from good observed hours only.

````python
bad_hum = df.hum_pct.eq(0)
df['hum_filled'] = df.get('hum_filled', False) | bad_hum
# typical humidity per (month, weather type), computed only from good, observed hours
good = df[~df.hum_filled & ~filled]
hum_med = good.groupby(['mnth', 'weathersit'])['hum_pct'].median()

# look up the median for each bad row's (month, weathersit) pair
keys = pd.MultiIndex.from_frame(df.loc[bad_hum, ['mnth', 'weathersit']])
df.loc[bad_hum, 'hum_pct'] = hum_med.reindex(keys).values
# keep the normalised column consistent
df.loc[bad_hum, 'hum'] = df.loc[bad_hum, 'hum_pct'] / 100

print('0 % hours left:', (df.hum_pct == 0).sum(), '| hours fixed:', df.hum_filled.sum())
````

## Stuck atemp on 2012-08-17 and 2012-12-29

We assume atemp is highly correlated with temp so we predict the stuck hours from temp with a straight line fitted on good hours (Lecture 2 – Slide 31: predictive model). Unlike a monthly median, this keeps the hour-by-hour shape of the day. The stuck days are taken straight from the Pass A table.

````python
# flagged days where atemp never changes while temp does -> stuck atemp sensor
stuck_days = top.index[(top.atemp_c_std == 0) & (top.temp_c_std > 0)]
print('stuck days:', list(stuck_days.date))

stuck = df.dteday.isin(stuck_days)
df['atemp_filled'] = df.get('atemp_filled', False) | stuck

# atemp follows temp closely, so fit a straight line on the good hours
# (not the hours we filled ourselves, and not the stuck hours)
good = ~filled & ~df.atemp_filled
coef = np.polyfit(df.loc[good, 'temp_c'], df.loc[good, 'atemp_c'], 1)
print(f"atemp ≈ {coef[0]:.2f} × temp {coef[1]:+.2f} | corr on good hours: "
      f"{df.loc[good, 'temp_c'].corr(df.loc[good, 'atemp_c']):.3f}")

df.loc[df.atemp_filled, 'atemp_c'] = np.polyval(coef, df.loc[df.atemp_filled, 'temp_c'])
# keep the normalised column consistent
df.loc[df.atemp_filled, 'atemp'] = (df.loc[df.atemp_filled, 'atemp_c'] + 16) / 66

df.loc['2012-08-17', ['hr', 'temp_c', 'atemp_c', 'atemp_filled']].head(4)
````

# Task 3: Dimension reduction and visualization

We use **PCA** for two things:

1. Visualization:
    Each day has 24 hourly rider values. Using PCA we reduces them to 2 so every day becomes one point on a plot (a scatterplot, Lecture 2 – Slide 56; looking at the data in a low-dimensional space shows its natural groups, Lecture 3 – Slide 113).

3. Confirming the rider difference from Task 2:
    Here we only use the 'cnt' value so PCA never sees the rider types.


### Choices and reasoning

- **One row per day, one column per hour.** We want to compare days, so each day is one row with its 24 hourly ride counts. This is the usual samples × features table (Lecture 1 – Slide 61). We did the same in Lab 4 – Task 5, where every day became one row, and in Lab 4 – Task 6 we also used 24 h as the natural cycle of the data.
- **Only demand goes in, not every column.** Adding weather, calendar codes and flags would put columns with different units and meanings in one table and mix the causes with the outcome (Lecture 1 – Slide 22: *"Feature selection is key!"*; Lecture 5 – Slide 16: irrelevant dimensions make distances meaningless).
- **Each day is divided by its own total.** Each value then becomes the hour's share of the day's rides, so PCA compares the shape of the day and not how busy it was. In Lab 4 – Task 5 we normalised every day for the same reason, *"so the clusters are about the shape and not the level"*. It is the level × pattern idea of the multiplicative decomposition (Lecture 6 – Slide 27).
- **No standardisation.** After the division all 24 columns are in the same unit (share of the day), which is what PCA needs (Lecture 5 – Slide 22: *"Features must have the same scale"*). Standardising on top would give the quiet night hours, which barely change, the same weight as the rush hours.
- **The 8 storm-shutdown days are left out.** Their outage hours have no ride count, and PCA *"cannot be used with missing data"* (Lecture 5 – Slide 22).
- **Number of components.** We look at the explained variance per component, as in Lab 3 – Task 3, before choosing.

````python
from sklearn.decomposition import PCA

# leave out the 8 storm-shutdown days: their outage hours have no ride count (NaN) and PCA can't use NaN
keep = ~df.outage.groupby(df.dteday).any()
# one row per day, one column per hour
rides = df.pivot(index='dteday', columns='hr', values='cnt')[keep]
# share of the day's rides in each hour: the shape of the day, not its volume
shape = rides.div(rides.sum(axis=1), axis=0)
# how many components: explained variance per component, as in Lab 3
pca_full = PCA().fit(shape)
ev = pca_full.explained_variance_ratio_
cum = np.cumsum(ev)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(range(1, 11), cum[:10], 'o-', color='black', label='cumulative')
ax.axhline(0.90, color='red', ls='--', lw=1)
ax.set(xlabel='component', ylabel='explained variance', xticks=range(1, 11),
       title='Explained variance of the day-shape PCA')
ax.legend()
savefig('pca_explained_variance')
plt.show()
print('each component:', np.round(ev[:5], 3))
print('cumulative:', np.round(cum[:5], 3), '| components for 90 %:', np.argmax(cum >= 0.90) + 1)
````

````python
pca = PCA(n_components=2)
Z = pca.fit_transform(shape)
print(f'days used: {len(shape)} of {len(keep)}')
print('explained variance: PC1 = {:.0%}, PC2 = {:.0%}'.format(*pca.explained_variance_ratio_))
print(f'correlation of PC1 with the casual share of the day: {np.corrcoef(Z[:, 0], daily.casual_share[keep])[0, 1]:.2f}')

working = daily.workingday[keep].values == 1
fig, axes = plt.subplots()
axes.scatter(Z[working, 0], Z[working, 1], s=10, color=REG, label='working day')
axes.scatter(Z[~working, 0], Z[~working, 1], s=10, color=CAS, label='weekend / holiday')
axes.set(title='Each day = one point', xlabel='PC1', ylabel='PC2')
axes.legend()
plt.tight_layout()
savefig('pca_days')
plt.show()
````

### Our interpretation

**Component Selection:** 
    One component explains 73.8 % of the variance, two explain 84.5 %, then 89.0 % and 91.3 %. The 90 % target we used in Lab 3 – Task 3 would need 4 components, but there PCA was the input to kNN, so it had to keep most of the information. Here PCA is only for looking at the data. After PC2 each extra component adds little (4.5 %, then 2.4 %: an elbow, Lecture 3 – Slide 87), and low-variance axes *"perhaps represent less relevant features"* (Lecture 5 – Slide 20). Two is also the most we can plot. So we keep 2.

**Plot Interpretation:**
- PC1 alone explains 74 % of how the shape of the day varies, so most of the difference between days lies along one axis.
- Working days and weekends/holidays form two separate groups along PC1, found without any labels, like the weekday cluster in Lab 4 – Task 5 (Lecture 3 – Slide 113).
- PC1 correlates −0.74 with the casual share of the day (correlation shows how two variables relate, Lecture 2 – Slide 34). PCA never saw the split into registered and casual riders, only the total, yet its main axis follows the share of casual riders. This confirms the Task 2 finding (the commuter vs leisure difference) with an independent method.

**Limitations:**
- PCA is a linear method (Lecture 5 – Slide 22), so it only finds straight-line combinations of the hours.
- PC2 explains only 11 %, so we don't base any conclusion on it.

````python
# daily rides over the two years with a 30-day rolling mean to show growth and season
roll = daily[['registered', 'casual']].rolling(30, center=True).mean()
fig, ax = plt.subplots(figsize=(13, 4))
ax.plot(daily.index, daily.registered, color=REG, alpha=0.25, lw=0.8)
ax.plot(daily.index, daily.casual, color=CAS, alpha=0.25, lw=0.8)
ax.plot(roll.index, roll.registered, color=REG, lw=2, label='registered (30-day mean)')
ax.plot(roll.index, roll.casual, color=CAS, lw=2, label='casual (30-day mean)')
ax.set(title='Rides per day, 2011-2012', ylabel='rides / day')
ax.legend()
savefig('rides_per_day')
plt.show()
````

The line chart shows what PCA leaves out on purpose, because PCA only looks at the shape of a day, not its volume:

# Task 5: Feature engineering

The data only relays the weather now but our question is about what the weather was and how long ago the rain stopped and how many riders there would have been without it. To answer that we add:

| Feature | Representation | Reasoning |
|---|---|---|
| `rain` | `weathersit >= 3`: light rain/snow, heavy rain/snow and the storm shutdowns | the event we study  |
| `hours_after_rain` | hours since the last rain hour: 0 = first dry hour, NaN while it rains and before the first rain | the recovery clock |
| `normal_registered`, `normal_casual` | kNN estimate of the riders an hour would have had without rain | the baseline (*"Level: the expected values or baseline"*, Lecture 6 – Slide 16) |
| `median_registered`, `median_casual` | baseline generated through a median table | independent robustness check |


````python
df['rain'] = df['weathersit'] >= 3
hours_after = []
counter = np.nan
# we are using the rows itself as a unit of time (1 hour)
# so we don't have to keep calculating using datetime
for is_rain in df['rain']:
    if is_rain:
        counter = -1
        hours_after.append(np.nan)
    else:
        counter += 1
        hours_after.append(counter)
df['hours_after_rain'] = hours_after
print('rain/snow hours:', df.rain.sum())
````

````python
# rain hours in the 7-hour window around each hour
rain_nearby = df['rain'].rolling(7, center=True).sum()
short_rain = df['rain'] & (rain_nearby == 1) & df['hr'].between(10, 17)
print('short daytime rains found:', short_rain.sum())

example = df.index[short_rain][0] 
df.loc[example - pd.Timedelta('3h'): example + pd.Timedelta('3h'), ['weather', 'rain', 'hours_after_rain']]
````

### Our interpretation

- There are 1,526 rain hours (weathersit 3 and 4).
- The hours before the very first rain in the data stay NaN, so they never get a made-up value.
- Because weathersit 4 counts as rain, the 89 outage hours are rain hours, the clock only starts once the system has reopened and the rain has stopped.

# Now we predict the "normal"(baseline) demand 

Basically what would be the normal demand would be at day if there was no rain. "Normal" / Baseline is the demand an hour would have had without rain, the level of the series (Lecture 6 – Slide 16). We predict it with kNN. For every hour, take the 6 nearest clean days (in time) with the same day type and hour, and average them. A clean hour has no rain in the previous 24 h and no shutdown, so the model never sees rain or its after-effect. We first check on held-out days that it really predicts normal demand.



### Choices and reasoning

**We only score daytime hours (07–21) because at night the normal is close to 0.**

- **Defining "clean" hour:**

    An hour to be clean it must not be raining, should not be within 24 h after rain and not a shutdown. The normal is learned only from clean hours so that it never sees rain or its after-effect.


- **k-NN regression as our primary prediction method for normal:**

    The kNN idea from Lecture 4 – Slide 53 we used it to average the neighbours' rides instead of scoring their distance. We predict the normal of an hour using the average of the 6 nearest clean days in time with the same day type (working day or not) and the same hour.


- **Nearest in time, same day type, same hour.**

    Demand has a trend (more rides in 2012 than in 2011) and a season (summer vs winter) as the line chart in Task 3 showed. Lecture 6 – Slide 20 also notes a "more reliable hypothesis due to temporal closeness". Days close in time are at the same level and in the same season and the same day type and hour remove the daily and weekly rhythm from Task 2 so what left in the value is the weather effect.

- **k = 6.**

    Enough days to average out one unusual day, few enough to stay in the same season. This doesn't matter much cause we ran the notebook with multiple k values [1,..15] and 1-12 gives the same conclusion of 3h / 9h.


- **The median table as our secondary prediction method for normal:**

    We use the median of the clean hours with the same year, month, day type and hour as our prediction for that particular hour. It is the same grouping as Pass B and needs no model. We keep kNN as the primary normal because it is an average. In Pass B we used the median because its groups contain the anomalies we were looking for and a median is not pulled by a few extreme days. Here the clean hours already leaves out rain, its after-effect and the shutdowns. If both normals give the same answer in Task 4, the result does not depend on how "normal" is built.


- **Verification method for our prediction model:**

  We hide random 2-day blocks (about one rain episode plus the 24 h after it) and predict them from the other clean days and compare with the real rides and repeated with 8 random draws. In Lab 4 – Task 4 we tested a forecast on the last 90 days because  forecast may only use the past (Lecture 6 – Slide 51 also splits into train and test) but the normal is not a forecast as it fills in days inside the series using the days around them so we hide days inside the series (like in MaskedAutoEncoder). 


- **Error measure:**

    The mean absolute error (MAE, as in Lab 4 – Task 4) as a percentage of the mean actual rides so registered and casual riders can be compared. We also report the bias (total predicted ÷ total actual − 1) because the recovery curve in Task 4 divides sums so a normal that is too high overall pulls the whole curve down.

````python
from sklearn.neighbors import KNeighborsRegressor

# clean = not raining, not within 24 h after rain, and not a shutdown
clean = ~df['rain'] & ~(df['hours_after_rain'] <= 24) & ~df['outage']
# days since 1 Jan 2011, so kNN can find the nearest days in time
df['day_no'] = (df['dteday'] - df['dteday'].min()).dt.days

def knn_normal(clean_d, rider, k=6):
    normal = pd.Series(np.nan, index=df.index)
    for (wd, hr), slot in df.groupby(['workingday', 'hr']):
        train = slot[clean_d[slot.index]]
        model = KNeighborsRegressor(n_neighbors=k).fit(train[['day_no']], train[rider])
        normal[slot.index] = model.predict(slot[['day_no']])
    return normal

def median_normal(clean_d, rider):
    keys = ['yr', 'mnth', 'workingday', 'hr']
    med = df[clean_d].groupby(keys)[rider].median()
    return pd.Series(med.reindex(pd.MultiIndex.from_frame(df[keys])).values, index=df.index)

for rider in ['registered', 'casual']:
    df['normal_' + rider] = knn_normal(clean, rider)
    df['median_' + rider] = median_normal(clean, rider)

no_median = df[df.median_registered.isna()]
print('clean hours:', clean.sum())
print('hours without a median normal:', len(no_median), '| days:', [str(d) for d in sorted(no_median.dteday.dt.date.unique())],
      '| hours of day:', [int(h) for h in sorted(no_median.hr.unique())], '| working day:', [int(w) for w in no_median.workingday.unique()])
````

````python
# check both normals on hours they did not see and hide random 2-day blocks (about one rain episode
# plus the 24 h after it), predict them from the other clean days, repeat with 8 random draws
all_days = df['dteday'].unique()
rows = []
for seed in range(8):
    rng = np.random.default_rng(seed)
    starts = rng.choice(np.arange(0, len(all_days) - 1, 2), size=40, replace=False)   # 40 blocks = 80 days
    hide = df['dteday'].isin(set(all_days[starts]) | set(all_days[starts + 1]))
    check = clean & hide & df.hr.between(7, 21)
    for name, method in [('kNN', knn_normal), ('median table', median_normal)]:
        for rider in ['registered', 'casual']:
            p, y = method(clean & ~hide, rider)[check], df.loc[check, rider]
            ok = p.notna()    # the median table has no value for a few hours
            rows.append({'normal': name, 'rider': rider,
                         'error %': 100 * np.abs(p[ok] - y[ok]).mean() / y[ok].mean(),
                         'bias %': 100 * (p[ok].sum() / y[ok].sum() - 1)})
print('average over 8 draws:')
pd.DataFrame(rows).groupby(['normal', 'rider']).mean().round(1)
````

### Our interpretation

| Normal | Registered: error / bias | Casual: error / bias |
|---|---|---|
| kNN | 14.7 % / +0.1 % | 28.6 % / +4.8 % |
| median table | 15.4 % / +0.1 % | 28.7 % / −0.9 % |

- **Both normals are about equally accurate** on hours they did not see (the reliability check of Lecture 7 – Slide 11): in a single daytime hour they are off by about 15 % for registered and about 29 % for casual riders. Casual riders are harder to predict because their demand depends more on the day (Task 2: casual riders are anywhere from 1 % to 51 % of a day's rides).
- **For registered riders neither is biased.** For casual riders the kNN normal is about 5 % too high overall, while the median table is about right. A normal that is too high makes the recovery look slower, so in Task 4 the median normal shows whether this matters for the answer.
- **The median table has no value for 48 hours:** the weekend afternoons (16:00–21:00) of August 2011, where every hour had rain within the 24 h before, so no clean hour exists for that group. When we use the median normal, these hours are left out.

````python
# one rain episode as an example
day = df.loc['2012-09-06'].set_index('hr')
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for ax, rider, color in [(axes[0], 'registered', REG), (axes[1], 'casual', CAS)]:
    for h in day.index[day.rain]:
        ax.axvspan(h - 0.5, h + 0.5, color='#e5e4e0', lw=0)      # rain hours shaded
    ax.plot(day.index, day[rider], color=color, lw=2, label='actual')
    ax.plot(day.index, day['normal_' + rider], color=color, lw=2, ls='-', alpha=0.6, label='normal (kNN)')
    ax.plot(day.index, day['median_' + rider], color=color, lw=2, ls='--', alpha=0.3, label='normal (median)')

    ax.set(title=f'{rider}: actual vs normal on 2012-09-06 (grey = rain)', xlabel='hour of day', xticks=range(0, 24, 3))
    ax.grid(axis='y', color='#e5e4e0', linewidth=0.8)
    ax.set_axisbelow(True)
    for side in ('top', 'right'):
        ax.spines[side].set_visible(False)
    ax.legend(loc='upper left')
axes[0].set_ylabel('rides / hour')
plt.tight_layout()
savefig('example_rain_day')
plt.show()

print('rain hours that day:', [int(h) for h in day.index[day.rain]])
first, dry = day.loc[12:14], day.loc[12:21]
for rider in ['registered', 'casual']:
    print(f"{rider:10s} actual ÷ normal knn | first 3 dry hours (12-14 h): {first[rider].sum() / first['normal_' + rider].sum():.2f}"
          f" | 12-21 h: {dry[rider].sum() / dry['normal_' + rider].sum():.2f}")
    
    print(f"{rider:10s} actual ÷ normal median | first 3 dry hours (12-14 h): {first[rider].sum() / first['median_' + rider].sum():.2f}"
          f" | 12-21 h: {dry[rider].sum() / dry['median_' + rider].sum():.2f}")
````

### Our interpretation

It rains until 11:00 and in the first three dry hours registered riders are at 65 % of normal and casual riders at 40 %. Over the dry afternoon (12–21 h) registered riders are back at 95 % of normal, while casual riders are still at 59 %.

This is a single day, so it only illustrates the idea. Task 4 measures the same thing over every rain episode in the two years.

# Research question

> **When rain or snow stops, how many hours does it take for bike demand to return to normal, and do registered riders (commuters) and casual riders recover at different speeds?**

### Choices and reasoning

- **Daytime hours only (07–21).** At night the normal is close to 0 rides, so ratios are mostly noise, and the company does not move bikes at 3 a.m.
- **The dry hours 0–24 h after rain**, grouped by `hours_after_rain`. Outage hours have no ride count and drop out.
- **Sum of actual ÷ sum of normal for each hour after rain**, not the mean of the hourly ratios. A quiet hour with 2 normal rides and 6 actual rides has a ratio of 3 and would pull a mean up; dividing the sums weights every hour by how busy it normally is. The ratio is observed ÷ expected, the same idea as lift (Lecture 7 – Slides 34–35; Lab 5 – Task 3).
- **Recovered = the ratio reaches 0.9 and stays there** (within 10 % of normal). The kNN normal itself is about 5 % too high for casual riders (Task 5 check), so a much stricter line would measure the normal's own error instead of the recovery.
- **Registered and casual riders separately**, because Task 2 and Task 3 showed they behave like two different groups.

````python
# dry daytime hours with a real ride count (outage hours have NaN and drop out)
daytime = df[df.hr.between(7, 21) & df.cnt.notna()]
after = daytime[daytime.hours_after_rain <= 24]

def recovery_curve(data, normal='normal_'):
    """actual ÷ normal rides for each hour after rain (sums, so a quiet hour can't blow up the ratio)"""
    g = data.groupby('hours_after_rain')
    return pd.DataFrame({'hours used': g.size(),
                         'registered': g.registered.sum() / g[normal + 'registered'].sum(),
                         'casual': g.casual.sum() / g[normal + 'casual'].sum()})

def recovery_hour(ratio, threshold=0.9):
    """first hour from which the ratio never drops below the threshold again (within 0-24 h)"""
    for h in ratio.index:
        if (ratio.loc[h:] >= threshold).all():
            return int(h)
    return None

curve = recovery_curve(after)
during = daytime[daytime.rain]
print(f'while it rains: registered at {during.registered.sum() / during.normal_registered.sum():.0%} of normal, '
      f'casual at {during.casual.sum() / during.normal_casual.sum():.0%}')
curve.T.round(2)
````

````python
rec = {r: recovery_hour(curve[r]) for r in ['registered', 'casual']}

fig, ax = plt.subplots(figsize=(11, 4.5))
ax.axhline(1.0, color='#52514e', lw=0.8)
ax.axhline(0.9, color='#a3a29d', lw=0.8)
ax.text(24.4, 0.9, '90 % of normal', va='center', fontsize=9, color='#52514e')
for rider, color in [('registered', REG), ('casual', CAS)]:
    ax.plot(curve.index, curve[rider], color=color, marker='o', ms=5, lw=2, label=rider)
    ax.axvline(rec[rider], color=color, lw=1, alpha=0.6)
    ax.text(rec[rider] + 0.2, 0.47, f'{rider}\nback at {rec[rider]} h', fontsize=9, color='#52514e')
ax.set(title='Demand after rain/snow stops (daytime hours, 2011-2012)', xlabel='hours since the rain/snow stopped',
       ylabel='actual ÷ normal rides', xticks=range(0, 25, 2), ylim=(0.45, 1.1))
ax.grid(axis='y', color='#e5e4e0', linewidth=0.8)
ax.set_axisbelow(True)
for side in ('top', 'right'):
    ax.spines[side].set_visible(False)
ax.legend(loc='lower right')
savefig('recovery_curve')
plt.show()
print('recovery time (h):', rec)
````

### Our interpretation

| | Registered (commuters) | Casual (leisure) |
|---|---|---|
| while it rains | 56 % of normal | 42 % of normal |
| first dry hour | 71 % | 54 % |
| back within 10 % of normal | after **3 h** | after **9 h** |

- **Demand does not come back the moment the rain stops.** In the first dry hour both groups are still clearly below normal, a surprising result if one expects riders to return as soon as it is dry.
- **Registered riders recover in 3 h, casual riders take 9 h.** Casual riders also drop more while it rains and start lower in the first dry hour.
- Every point on the curve is based on more than 100 daytime hours (row `hours used`), so the curve is not driven by a few hours.

## Using second baseline

The answer depends on the baseline so we repeat the curve with the median-table normal from Task 5. The median is the *typical* day and is not pulled by a few very busy days the kNN normal is an *average* of the nearest clean days (kNN, Lecture 4 – Slide 53). If both give the same answer, the result does not depend on how baseline is built.

````python
# robustness check: 
# the same curve with the median normal (hours without a median normal are left out)
with_median = after[after.median_registered.notna()]
curve_median = recovery_curve(with_median, normal='median_')
rec_median = {r: recovery_hour(curve_median[r]) for r in ['registered', 'casual']}
print('hours left out (no median normal):', len(after) - len(with_median))
print('recovery time (h): kNN normal', rec, '| median normal', rec_median)

fig, axes = plt.subplots(1, 2, figsize=(13, 4), sharey=True)
for ax, rider, color in [(axes[0], 'registered', REG), (axes[1], 'casual', CAS)]:
    ax.axhline(0.9, color='#a3a29d', lw=0.8)
    ax.plot(curve.index, curve[rider], color=color, marker='o', ms=5, lw=2, label='kNN normal')
    ax.plot(curve_median.index, curve_median[rider], color=color, lw=2, ls='--', label='median normal')
    ax.set(title=f'{rider}: back at {rec[rider]} h (kNN) and {rec_median[rider]} h (median)',
           xlabel='hours since the rain/snow stopped', xticks=range(0, 25, 4))
    ax.grid(axis='y', color='#e5e4e0', linewidth=0.8)
    ax.set_axisbelow(True)
    for side in ('top', 'right'):
        ax.spines[side].set_visible(False)
    ax.legend(loc='lower right')
axes[0].set_ylabel('actual ÷ normal rides')
plt.tight_layout()
savefig('recovery_both_normals')
plt.show()
````

### Our interpretation

- **Registered riders are back after 3 h with both normals.**
- **Casual riders are back after 9 h with the kNN normal and 7 h with the median normal.** The kNN normal for casual riders is about 5 % too high (Task 5 check: bias +4.8 % against −0.9 % for the median table) because a few very busy days pull an average up. Measured against a normal that is too high, recovery looks slower.
- **So the answer is 3 h against 7–9 h:** whichever normal we use, casual riders take more than twice as long as registered riders.


## What it means for the company

To turn the curve into something the company can use, we count the **rides lost**: normal minus actual rides in the daytime hours, while it rains and in the 24 h after, per year (the data covers two years, so we divide by 2).

````python
lost = {}
for rider in ['registered', 'casual']:
    gap = daytime['normal_' + rider] - daytime[rider]          
    lost[rider] = {'while raining': gap[daytime.rain].sum() / 2,
                   'after: 0-4 h': gap[daytime.hours_after_rain.between(0, 4)].sum() / 2,
                   'after: 5-24 h': gap[daytime.hours_after_rain.between(5, 24)].sum() / 2}
lost = pd.DataFrame(lost)
lost['total'] = lost.sum(axis=1)
lost.loc['all weather losses'] = lost.sum()
lost.loc['share lost AFTER the rain'] = (lost.loc['after: 0-4 h'] + lost.loc['after: 5-24 h']) / lost.loc['all weather losses']
lost.round(2)
````

````python
# where the weather losses happen, per rider type
periods = ['while raining', 'after: 0-4 h', 'after: 5-24 h']
x = np.arange(len(periods))
w = 0.38
fig, ax = plt.subplots(figsize=(9, 4))
for i, (rider, color) in enumerate([('registered', REG), ('casual', CAS)]):
    vals = lost.loc[periods, rider]
    bars = ax.bar(x + (i - 0.5) * w, vals, width=w - 0.03, color=color, label=rider)
    ax.bar_label(bars, labels=[f'{v / 1000:.1f}k' for v in vals], padding=2, fontsize=9, color='#52514e')
ax.axhline(0, color='#52514e', lw=0.8)
ax.set(xticks=x, xticklabels=periods, ylabel='daytime rides lost per year', title='Where the weather losses happen')
ax.grid(axis='y', color='#e5e4e0', linewidth=0.8)
ax.set_axisbelow(True)
for side in ('top', 'right'):
    ax.spines[side].set_visible(False)
ax.legend()
savefig('rides_lost')
plt.show()
````

### Our interpretation

- **Bad weather costs about 101,000 daytime rides a year**, and **39 %** of them are lost *after* the rain has stopped.
- **Registered riders:** 33 % of their weather loss comes after the rain, almost all of it in the first 6 h (the 7–24 h bar is about 0: they are back to normal).
- **Casual riders:** 52 %, so more than half of their weather loss comes after the rain, and it is still there 7–24 h later.

**Recommendations** 
1. **Forecast with a recovery clock.** Add `hours_after_rain` to the demand forecast. Without it, the hours after rain are over-predicted (short-term forecasting, Lecture 6 – Slide 8).
2. **Time the rebalancing.** Commuter stations need their normal number of bikes about 3 h after the rain stops, not the moment it stops. Leisure stations stay quiet for 7–9 h, which is free truck time for moving bikes and for maintenance.
3. **Win back casual riders.** Half of their weather loss comes after the rain. A message like "the sun is out: 24-h pass discount" targets exactly that window.

**Limitations**
- We count rides not revenue.
- "Rain" (`weathersit >= 3`) mixes light rain, light snow and heavy storms, and the weathersit 4 of the outage hours comes from the storm reports, not from a measurement.
- Only daytime hours are counted, and the data covers one city and two years.
- We cannot separate the rain from what comes with it afterwards (for example humid air or muddy enviroment). For the company that matters little, because the low demand in those hours is real either way.
